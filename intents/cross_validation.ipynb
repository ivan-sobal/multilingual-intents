```python
import numpy as np
from sklearn.model_selection import StratifiedKFold, cross_val_score

from intents.baseline import build_pipeline, evaluate
from intents.data import load_split
```

```python
texts, labels = load_split("en", "train")
texts = np.array(texts)
labels = np.array(labels)
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

```python

scores = []
for train_index, test_index in skf.split(texts, labels):
    model = build_pipeline(analyzer="char_wb", ngram_min=2, ngram_max=5, C=30)
    model.fit(texts[train_index], labels[train_index])
    scores.append(evaluate(model, texts[test_index], labels[test_index]))
    print(f"fold {len(scores)}: {scores[-1]:.4f}")

print(f"mean {np.mean(scores):.4f}, std {np.std(scores):.4f}")

cross_val_score(build_pipeline(analyzer="char_wb", ngram_min=2, ngram_max=5, C=30), texts, labels, cv=skf)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
model = build_pipeline(analyzer="char_wb", ngram_min=2, ngram_max=5)
param_grid = {"logisticregression__C": [1, 3, 10, 30, 100]}
search = GridSearchCV(model, param_grid, cv=skf, scoring="accuracy",n_jobs=-1)
search.fit(texts, labels)
print(search.best_params_)
print(search.best_score_)
print(search.cv_results_["mean_test_score"], search.cv_results_["std_test_score"])
```
