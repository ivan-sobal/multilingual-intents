```python
from collections import Counter

import pandas as pd
from sklearn.metrics import classification_report

from intents.baseline import build_pipeline
from intents.data import load_split

train_texts, train_labels = load_split("en", "train")
val_texts, val_labels = load_split("en", "validation")

model = build_pipeline()
model.fit(train_texts, train_labels)
preds = model.predict(val_texts)
```

```python
report = classification_report(val_labels, preds, output_dict=True, zero_division=0)
df = pd.DataFrame(report).T
df = df.drop(["accuracy", "macro avg", "weighted avg"])
df.sort_values(["f1-score"]).head(3)  # the 3 intents with the lowest F1
(preds == "cooking_query").sum()
```
